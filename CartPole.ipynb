```python
import os
import gymnasium as gym
from stable_baselines3 import PPO
from stable_baselines3.common.vec_env import DummyVecEnv
from stable_baselines3.common.evaluation import evaluate_policy
```

```python
environment_name = 'CartPole-v0'
env = gym.make(environment_name)
```

```python
#Load environment
episodes = 5
for episode in range(1, episodes+1):
    state = env.reset()
    done = False
    score = 0
    
    while not done:
        env.render()
        action = env.action_space.sample()
        n_state, reward, done, trunc, info = env.step(action)
        score += reward
    print('Episode:{} Score:{}'.format(episode, score))
env.close()
```

```python
#Train RL Model
log_path = os.path.join('Training', 'Logs')
```

```python
env = DummyVecEnv([lambda: gym.make(environment_name)])
model = PPO('MlpPolicy', env, verbose = 1, tensorboard_log=log_path)
```

```python
model.learn(total_timesteps=20000)
```

```python
env.close()
```

```python
#Save and Reload
PPO_Path = os.path.join('Training', 'Saved Models', 'PPO_Model_Cartpole1')
```

```python
model.save(PPO_Path)
```

```python
del model
```

```python
model.learn(total_timesteps=1000)
```

```python
model = PPO.load(PPO_Path, env=env)
```

```python
env = gym.make(environment_name, render_mode="human")
```

```python
#Evaluation
evaluate_policy(model, env, n_eval_episodes=10, render=True)
```

```python
env.close()
```

Adding a callback to the training stage

```python
from stable_baselines3.common.callbacks import EvalCallback, StopTrainingOnRewardThreshold
```

```python
save_path = os.path.join('Training', 'Saved Models')
```

```python
stop_callback = StopTrainingOnRewardThreshold(reward_threshold=200, verbose=1)
eval_callback = EvalCallback(env, callback_on_new_best=stop_callback,eval_freq=10000, best_model_save_path=save_path, verbose=1)
```

```python
model = PPO('MlpPolicy', env, verbose = 1, tensorboard_log=log_path)
```

```python
model.learn(total_timesteps=20000, callback = eval_callback)
```

Changing Policies

```python
net_arch = [dict(pi =[128,128,128,128], vf=[128,128,128,128])]
```

```python
model = PPO('MlpPolicy', env, verbose=1, tensorboard_log=log_path, policy_kwargs={'net_arch':net_arch})
```

```python
model.learn(total_timesteps=20000, callback = eval_callback)
```
